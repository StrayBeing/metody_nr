```python
import numpy as np

def suma(f, xp, xk, k):
    res = 0
    x = xp
    while x <= xk:
        value = f(x)
        if value is None:
            res = 0
            return res
        x += k
    return res

def iloczyn(f, xp, xk, k):
    res = 1
    x = xp
    while x <= xk:
        value = f(x)
        if value is not None:
            res *= value
        x += k
    return res

def oneX(x):
    try:
        return 1 / x
    except ZeroDivisionError:
        print("DZIELENIE PRZEZ ZERO")
        return None

def sinus(x):
    return np.sin(x)

if __name__ == "__main__":
    pocz = 0
    kon = 3.14
    kro = 0.01
    suma_result = suma(oneX, pocz, kon, kro)
    print("Suma 1/x dla zakresu %f do %f o kroku %f to: %f" % (pocz, kon, kro, suma(oneX, pocz, kon, kro)))
    print("Suma sin(x) dla zakresu %f do %f o kroku %f to: %f" % (pocz, kon, kro, suma(sinus, pocz, kon, kro)))

    iloczyn_result = iloczyn(oneX, pocz, kon, kro)
    if iloczyn_result is not None:
        print("Iloczyn 1/x dla zakresu %f do %f o kroku %f to: %f" % (pocz, kon, kro, iloczyn_result))
    else:
        print("Błąd w obliczeniach iloczynu 1/x.")

    iloczyn_result = iloczyn(sinus, pocz, kon, kro)
    if iloczyn_result is not None:
        print("Iloczyn sin(x) dla zakresu %f do %f o kroku %f to: %f" % (pocz, kon, kro, iloczyn_result))
    else:
        print("Błąd w obliczeniach iloczynu sin(x).")

```
