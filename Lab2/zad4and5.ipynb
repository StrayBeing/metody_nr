```python
import numpy as np
import matplotlib.pyplot as plt

def rysuj(f, xp, xk, k,opcja):
    x = np.arange(xp, xk + k, k)
    y = [f(i) for i in x]
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, label=f.__name__)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Wykres funkcji {f.__name__}')
    plt.legend()
    if opcja==1:
     plt.grid(True)
     plt.plot(x, y, marker='o', label=f.__name__)
    plt.show()
# Przykłady funkcji
def funkcja_kwadratowa(x):
    return x**2
def funkcja_liniowa2xplus5(x):
    return 2*x+5
if __name__ == "__main__":
    xp = -20
    xk = 20
    krok = 2
    rysuj(funkcja_liniowa2xplus5, xp, xk, krok,0)
    rysuj(funkcja_kwadratowa, xp, xk, krok,1)
```
