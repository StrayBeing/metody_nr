```python
import numpy as np

def f1(x):
    try:
      return 1 / x
    except ZeroDivisionError:
      return 1

def f2(x):
    return np.sin(x)

def suma(f, xp, xk, k):
    x = np.arange(xp, xk + k, k)
    return np.sum(f(x))

def iloczyn(f, xp, xk, k):
    x = np.arange(xp, xk + k, k)
    return np.prod(f(x))

print(f"Wynik sumy dla f(x) = 1/x: {suma(f1, -3.14, 3.14, 0.01)}")
print(f"Wynik iloczynu dla f(x) = 1/x: {iloczyn(f1, -3.14, 3.14, 0.01)}")
print(f"Wynik sumy dla f(x) = sin(x): {suma(f2, -3.14, 3.14, 0.01)}")
print(f"Wynik iloczynu dla f(x) = sin(x): {iloczyn(f2, -3.14, 3.14, 0.01)}")
print(f"f(0)={f1(0)}")
```
