```python
x=3.0
xn=0.1
blad=0.000001
x_old=0
while abs(x_old-xn)>blad:
	x_old=xn
	xn=(xn+x/xn)/2
print("x=%lf" % xn)
"""
Program zaprezentowany powyżej służy do obliczenia pierwiastka kwadratowego z liczby zapisanej w zmiennej x
Wykorzystuje on metodę Newtona-Raphsona. Zmienna xn jest to początkowa przybliżona wartość pierwiastka.
Zmienna bląd wskazuje na dopuszczalny błąd obliczeń.
x_old jest zmienna pomocnicza i służy do zapamiętania poprzedniej wartości pierwiastka.
Metoda ta stopniowo poprawia przybliżenie pierwiastka aż różnica między kolejnymi przybliżonymi wartościami stanie się mniejsza niż błąd.
tak więc petla while wykonuje się do momentu gdy różnica między obecną a poprzednią wartością przybliżną pierwiastka stanie sie mniejsza lub rowna zmiennej błąd
Na końcu wyświetlany jest wynik.
"""
```
